# Originalprogramm wiederherstellen

Dieses Rettungs-Notebook lädt die aktuelle unveränderte Masterversion aus dem GitHub-Repository herunter. Deine CSV-Dateien und bisherigen Notebooks werden **nicht gelöscht**.

**Anleitung:** Klicke in die nächste Zelle und drücke **Shift + Enter**. Öffne anschließend die neu erzeugte Datei im Dateibrowser links.

Voraussetzung: Internetverbindung und ein funktionierender **Python (Pyodide)**-Kernel.

```python
from datetime import datetime
from pathlib import Path
from pyodide.http import pyfetch
import json

MASTER_URL = (
    "https://raw.githubusercontent.com/"
    "SuperfroggyX/Katalase/main/content/01_Katalase_CSV_Graphen.ipynb"
)

response = await pyfetch(MASTER_URL)
if not response.ok:
    raise RuntimeError(f"Masterdatei konnte nicht geladen werden: HTTP {response.status}")

content = await response.string()
notebook = json.loads(content)
if notebook.get("nbformat") != 4 or not isinstance(notebook.get("cells"), list):
    raise ValueError("Die geladene Datei ist kein gültiges Jupyter-Notebook.")

timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
target = Path(f"Katalase_FRISCH_{timestamp}.ipynb")
counter = 1
while target.exists():
    target = Path(f"Katalase_FRISCH_{timestamp}_{counter}.ipynb")
    counter += 1
target.write_text(content, encoding="utf-8")
print(f"Fertig! Öffne links im Dateibrowser: {target.name}")
print("Deine bisherigen Notebooks und CSV-Dateien bleiben erhalten.")
```
