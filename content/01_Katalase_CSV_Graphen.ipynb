The graph displays foam height, an indirect indicator of catalase activity.


```python
import csv
import matplotlib.pyplot as plt

# ===== EDIT THESE SETTINGS =====
filename = "katalase_messwerte.csv"
x_column = "Time_min"
y_column = "Foam_height_mm"
graph_type = "scatter"  # "scatter", "line", or "bar"
colour = "blue"         # e.g. "red", "green", "orange"
x_min, x_max = 0, 5
y_min, y_max = 0, 35
show_grid = True

# ===== LOAD CSV (comma or semicolon separated) =====
with open(filename, encoding="utf-8-sig", newline="") as file:
    sample = file.read(2048)
    file.seek(0)
    dialect = csv.Sniffer().sniff(sample, delimiters=",;\t")
    reader = csv.DictReader(file, dialect=dialect)
    if x_column not in reader.fieldnames or y_column not in reader.fieldnames:
        raise ValueError(f"Columns available: {reader.fieldnames}")
    rows = list(reader)

# Decimal commas are supported in semicolon-separated files.
x = [float(row[x_column].strip().replace(",", ".")) for row in rows]
y = [float(row[y_column].strip().replace(",", ".")) for row in rows]
if not x:
    raise ValueError("The CSV contains no measurements.")

# ===== CREATE GRAPH =====
fig, ax = plt.subplots(figsize=(8, 5))
if graph_type == "scatter":
    ax.scatter(x, y, color=colour, s=60)
elif graph_type == "line":
    ax.plot(x, y, color=colour, marker="o")
elif graph_type == "bar":
    ax.bar(x, y, color=colour, width=0.6)
else:
    raise ValueError("graph_type must be scatter, line or bar")

ax.set_xlabel("Time (min)")
ax.set_ylabel("Foam height (mm)")
ax.set_title("Catalase experiment")
ax.set_xlim(x_min, x_max)
ax.set_ylim(y_min, y_max)
ax.grid(show_grid, alpha=0.3)
fig.tight_layout()
# Uncomment to save a PNG in the JupyterLite file browser:
# fig.savefig("catalase_graph.png", dpi=200)
plt.show()
```
